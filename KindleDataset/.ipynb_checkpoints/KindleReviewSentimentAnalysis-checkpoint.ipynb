### LABEL CREATION

```python
def sentiment_label(r):
    if r <= 2:
        return "negative"
    elif r == 3:
        return "neutral"
    else:
        return "positive"

df['sentiment'] = df['rating'].apply(sentiment_label)
```

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

# downloads (run once)
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text(text):
    # lowercase
    text = text.lower()
    
    # remove special chars
    text = re.sub(r'[^a-z\s]', '', text)
    
    # tokenize
    tokens = word_tokenize(text)
    
    # remove stopwords + lemmatize
    tokens = [
        lemmatizer.lemmatize(word) 
        for word in tokens 
        if word not in stop_words
    ]
    
    return " ".join(tokens)

df['clean'] = df['reviewText'].apply(clean_text)
```

### PART 1 — BAG OF WORDS MODEL

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df['clean'], df['sentiment'], test_size=0.2, random_state=42
)
```

```python
vectorizer = CountVectorizer(max_features=5000)

X_train_bow = vectorizer.fit_transform(X_train)
X_test_bow = vectorizer.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression(max_iter=200)
lr.fit(X_train_bow, y_train)
pred_lr = lr.predict(X_test_bow)
```

```python
from sklearn.naive_bayes import MultinomialNB

nb = MultinomialNB()
nb.fit(X_train_bow, y_train)
pred_nb = nb.predict(X_test_bow)
```

```python
from sklearn.svm import LinearSVC

svm = LinearSVC()
svm.fit(X_train_bow, y_train)
pred_svm = svm.predict(X_test_bow)
```

```python
from sklearn.metrics import classification_report, accuracy_score

print("Logistic Regression\n", classification_report(y_test, pred_lr))
print("Naive Bayes\n", classification_report(y_test, pred_nb))
print("SVM\n", classification_report(y_test, pred_svm))
```

### Accuracy Comparison Graph 

```python
import matplotlib.pyplot as plt

acc_lr = accuracy_score(y_test, pred_lr)
acc_nb = accuracy_score(y_test, pred_nb)
acc_svm = accuracy_score(y_test, pred_svm)

models = ['LogReg', 'NaiveBayes', 'SVM']
scores = [acc_lr, acc_nb, acc_svm]

plt.figure()
plt.bar(models, scores)
plt.title("Model Accuracy Comparison")
plt.ylabel("Accuracy")
plt.show()
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_test, pred_svm)
plt.title("Confusion Matrix (Best Model)")
plt.show()
```

```python
##Top Important Words

import numpy as np

feature_names = vectorizer.get_feature_names_out()

for i, label in enumerate(lr.classes_):
    coefs = lr.coef_[i]
    
    top_pos = np.argsort(coefs)[-10:]
    top_neg = np.argsort(coefs)[:10]
    
    print("\nCLASS:", label)
    print("Most important words:", feature_names[top_pos])
    print("Opposite words:", feature_names[top_neg])
```

```python
## GRAPH


import matplotlib.pyplot as plt

def plot_top_words(class_index, title):
    coefs = lr.coef_[class_index]
    top = np.argsort(coefs)[-10:]
    
    words = feature_names[top]
    scores = coefs[top]
    
    plt.figure()
    plt.barh(words, scores)
    plt.title(title)
    plt.show()

plot_top_words(0,"Negative words")
plot_top_words(1,"Neutral words")
plot_top_words(2,"Positive words")
```

```python
## Model Checking


samples = [

    input(str(""))
]

# clean all
samples_clean = [clean_text(s) for s in samples]

# vectorize
samples_vec = vectorizer.transform(samples_clean)

# predict
preds = lr.predict(samples_vec)

for s,p in zip(samples,preds):
    print("\nText:", s)
    print("Prediction:", p)
```

#### PART 3 — WORD2VEC MODEL (ADVANCED)

```python
from gensim.models import Word2Vec

sentences = [text.split() for text in df['clean']]

w2v = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4
)
```

```python
import numpy as np

def sentence_vector(text):
    words = text.split()
    vecs = []

    for w in words:
        if w in w2v.wv:
            vecs.append(w2v.wv[w])

    if len(vecs)==0:
        return np.zeros(100)

    return np.mean(vecs, axis=0)

X_w2v = np.array([sentence_vector(t) for t in df['clean']])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_w2v, df['sentiment'], test_size=0.2, random_state=42
)

model = LogisticRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)
print(classification_report(y_test, pred))
```
