```python
paragraph = """
Artificial Intelligence is transforming the world rapidly. 
Many companies are investing heavily in AI technologies, and people are using AI tools daily.
However, understanding how these systems are working and how they can be improved is important.
Researchers are developing smarter models, improving accuracy, and reducing computational cost.
In the future, AI will impact healthcare, education, business, and almost every industry.
"""
```

```python
text = paragraph.lower()
print(text)
```

```python
import string

text = text.translate(str.maketrans('', '', string.punctuation))
print(text)
```

```python
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text)
print(tokens)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered = [w for w in tokens if w not in stop_words]
print(filtered)
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet
from nltk import pos_tag

lemmatizer = WordNetLemmatizer()

def get_pos(tag):
    if tag.startswith('J'):
        return wordnet.ADJ
    elif tag.startswith('V'):
        return wordnet.VERB
    elif tag.startswith('N'):
        return wordnet.NOUN
    elif tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

tagged = pos_tag(filtered)

clean_words = []
for word, tag in tagged:
    clean_words.append(lemmatizer.lemmatize(word, get_pos(tag)))

print(clean_words)
```
